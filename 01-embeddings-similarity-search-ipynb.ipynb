```python
from sentence_transformers import SentenceTransformer, util
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
sentences = [
    "How old are you?",
    "I love cats",
    "But I only find them cute when they dont touch me.",
    "I love loded fries.",
    "I love white sauce pasta.",
    "I want good consistent routine",
    "Need to become very rich to help ALLAHs creation.",
    "Love swimming and going gym everyday.",
    "Want to travel the world.",
    "Praying and need to get rid of angr issues",
]
```

**Trial 1**

```python
question = "How early birds start chirping in morning?"

sent_emb = model.encode(sentences)      # sentences -> numbers
q_emb = model.encode(question)          # question -> numbers
scores = util.cos_sim(q_emb, sent_emb)[0]

top3 = scores.argsort(descending=True)[:3]
for i in top3:
    print(round(float(scores[i]), 2), "|", sentences[i])
```
